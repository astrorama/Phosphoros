```python
import numpy as np
import scipy.integrate as integrate
from astropy.table import Table
import matplotlib.pyplot as plt
```

```python
def computePitAndCrps(pdf_data, pdf_bins, reference_values, progress_callback=None):
    """Compute the Probability Integral Transform (PIT) 
    
    Parameters:
    
    pdf_data (array(array)): an array (1 element by object) of 1-PDF values(evaluated at the 'pdf-bins' point)
    reference_values (array): for each object the 'True' value of the PDF parameter
    
    Returns:
    array, array: list of PIT
    
    """
    total = len(reference_values)
    pits = np.zeros(total)
    crps = np.zeros(total)
    samples = np.dot(np.ones((pdf_data.shape[0],1)),np.reshape(pdf_bins,(1,pdf_data.shape[1])))
    cumul_data = np.zeros(pdf_data.shape)
    for index in range(1,pdf_data.shape[1]):
        cumul_data[:,index]= cumul_data[:,index-1]+0.5*(pdf_data[:,index-1]+pdf_data[:,index])/(samples[:,index]-samples[:,index-1])

    for index, ref_value in enumerate(reference_values):
        current_bin = np.array(pdf_bins)
        current_data = np.array(pdf_data[index])
        current_cumul_data = cumul_data[index]
        if not ref_value in current_bin:
            current_bin = np.concatenate((current_bin,[ref_value]))
            current_bin.sort()
            current_data = np.interp(current_bin, pdf_bins, current_data)
            current_cumul_data = np.interp(current_bin, pdf_bins, current_cumul_data)
        current_data = current_data / np.trapezoid(current_data, current_bin)
        index_cut = np.argwhere(current_bin==ref_value)[0][0]
        if index_cut > 0:
            pits[index] = np.trapezoid(current_data[:index_cut], current_bin[:index_cut]) 
            if index_cut < len(current_bin)-1:
                crps[index] = np.trapezoid(np.square(current_cumul_data[:index_cut]), current_bin[:index_cut]) + np.trapezoid(np.square(current_cumul_data[index_cut:] -1), current_bin[index_cut:])
            else:
                crps[index] = np.trapezoid(np.square(current_cumul_data), current_bin) 
        else:
            crps[index] = np.trapezoid(np.square(current_cumul_data -1), current_bin)
            
        if progress_callback is not None:
            progress_callback(index,total)
    return pits, crps   
```

```python
def plotPIT( pits, histo_bins=20,**kwargs):
    """Plot an histogram of the PIT values 
    
    Parameters:
    
    pits (array): set of PIT value
    histo_bins(int): Number of bins for the histogram

    """
    mean = 1
    plt.title("PDF's PIT")
    plt.hist(pits,bins=histo_bins,range=(0,1),density=True,**kwargs)
    plt.plot([0,1],[mean,mean])
```

```python
def plotCRPS( crps, histo_bins=20,**kwargs):
    """Plot an histogram of the CRPS values 
    
    Parameters:
    
    crps (array): set of PIT value
    histo_bins(int): Number of bins for the histogram

    """
    plt.title("PDF's CRPS")
    plt.hist(crps,bins=histo_bins,range=(0,crps.max()),density=True,**kwargs)
```

```python
pdf = Table.read('/home/dubathf/Phosphoros/Results/cosmos/COSMOS2020_ZCOSMOS_goodz_nostar_50/phz_cat.fits')['Z-1D-PDF']
```

```python
z_true = Table.read('/home/dubathf/Phosphoros/Catalogs/cosmos/COSMOS2020_ZCOSMOS_goodz_nostar_50.fits')['z_spec']
```

```python
sampling = np.array(range(pdf.shape[1]))/100.0
```

```python
pits, crps = computePitAndCrps(pdf,sampling, z_true)
plotPIT(pits)
```

```python
plotCRPS(crps)
```
